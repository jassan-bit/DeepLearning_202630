# E3 · Revisión bibliográfica de R1 y guardado en GitHub

Este cuaderno recoge seis consultas en inglés y español mediante cinco API bibliográficas. Usa **CPU y la biblioteca estándar de Python**; no instala paquetes, no usa GPU ni entrena modelos.

El candidato revisado es la cabeza aditiva positiva de **1210 parámetros**. La formulación histórica se conserva; la ejecución registra el commit exacto del código consultado.

## Uso
1. Usa una sesión de Colab con CPU.
2. Habilita el secreto de Colab `GITHUB_TOKEN` con acceso al repositorio y **Contents: Read and write**. Si no existe, el programa lo solicita mediante entrada oculta. No pegues el token en una celda.
3. Ejecuta la celda. Cada proveedor procesa seis consultas con pausas; algunos pueden rechazar solicitudes o exigir credenciales propias. Los fallos conservan conteos desconocidos.
4. El resultado se guarda en una rama nueva `e3-r1-search-...`, bajo `runs/r1_search/...`, con manifiesto de integridad. Conserva la sesión hasta ver la URL de GitHub.
5. Si falla el guardado, reejecuta la misma celda: conserva las consultas locales y reintenta guardarlas. Tras un reinicio, ese contexto local se pierde.

El recolector conserva la primera página y los conteos nativos; **no completa el cribado, la revisión de textos, el seguimiento bibliográfico ni el concepto humano externo**. El cumplimiento de R1 requiere evaluar la diferencia sustantiva y obtener la revisión real de una persona ajena al diseño. Tres bases que respondan no aprueban automáticamente R1.

Preparado con asistencia de IA. No contiene un concepto humano, firma o declaración de novedad acreditada.


```python
import json
import urllib.request

# Fija una sola versión del código; al reintentar conserva la ejecución local.
_context = globals().get("_E3_R1_SEARCH_CONTEXT")
if _context:
    R1_SOURCE_COMMIT = _context["source"]
else:
    _ref_url = "https://api.github.com/repos/jassan-bit/DeepLearning_202630/git/ref/heads/main"
    with urllib.request.urlopen(_ref_url, timeout=30) as _response:
        R1_SOURCE_COMMIT = json.load(_response)["object"]["sha"]
if len(R1_SOURCE_COMMIT) != 40 or any(c not in "0123456789abcdef" for c in R1_SOURCE_COMMIT):
    raise ValueError("No se obtuvo un commit válido")
_launcher_url = (
    "https://raw.githubusercontent.com/jassan-bit/DeepLearning_202630/"
    + R1_SOURCE_COMMIT
    + "/experiments/e3_r1_review_20261118/search/launch_colab.py"
)
print("Código fijado:", R1_SOURCE_COMMIT)
with urllib.request.urlopen(_launcher_url, timeout=30) as _response:
    _launcher_code = _response.read().decode("utf-8")
exec(compile(_launcher_code, _launcher_url, "exec"), globals())
```
